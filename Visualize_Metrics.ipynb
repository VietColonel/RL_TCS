# PressLight Performance Visualization
This notebook visualizes the learning curve (Average Travel Time) of the PressLight RL agents across 4 different synthetic traffic scenarios.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

base_dir = os.path.join("records", "test")
folders = {
    "Light-Flat (300, 0.3)": "anon_1_6_300_0.3_synthetic.json_09_28_00_40_21",
    "Light-Peak (300, 0.6)": "anon_1_6_300_0.6_synthetic.json_09_28_00_40_21",
    "Heavy-Flat (700, 0.3)": "anon_1_6_700_0.3_synthetic.json_09_28_00_40_21",
    "Heavy-Peak (700, 0.6)": "anon_1_6_700_0.6_synthetic.json_09_28_00_40_21"
}

dataframes = []
for label, folder in folders.items():
    csv_path = os.path.join(base_dir, folder, "metrics_log.csv")
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        df["Scenario"] = label
        dataframes.append(df)

if dataframes:
    full_df = pd.concat(dataframes, ignore_index=True)
    print("Data loaded successfully!")
else:
    print("No metrics_log.csv files found!")
```

```python
plt.figure(figsize=(14, 8))
ax = sns.lineplot(data=full_df, x="round", y="travel_time", hue="Scenario", linewidth=2)

plt.title("PressLight Convergence: Average Travel Time over 300 Rounds", fontsize=16, fontweight="bold", pad=20)
plt.xlabel("Training Round (Episode)", fontsize=14)
plt.ylabel("Average Travel Time (s)", fontsize=14)
plt.legend(title="Traffic Scenario", title_fontsize='13', fontsize='12')
plt.tight_layout()
plt.show()
```

### Minimum Travel Times Achieved
Below are the best (minimum) travel times achieved by the RL agent for each scenario during the 300 rounds.

```python
best_results = full_df.groupby("Scenario")["travel_time"].min().reset_index()
best_results = best_results.rename(columns={"travel_time": "Best Average Travel Time (s)"})
best_results.sort_values("Best Average Travel Time (s)")
best_results
```
